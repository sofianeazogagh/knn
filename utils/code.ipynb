```python
import numpy as np
import pandas as pd


def split_dataset(dataset_name, train_size, seed):
    np.random.seed(seed)
    dataset = pd.read_csv(f"../data/{dataset_name}.csv", header=None)
    train_dataset = dataset.sample(frac=train_size).reset_index(drop=True)
    test_dataset = dataset.drop(train_dataset.index)
    
    train_dataset.to_csv(f"../data/train/{dataset_name}_train.csv", index=False, header=False)
    test_dataset.to_csv(f"../data/test/{dataset_name}_test.csv", index=False, header=False)
    
    print("Number of training points: ", train_dataset.shape[0])
    print("Number of test points: ", test_dataset.shape[0])
    


split_dataset("cancer", 0.95, 42)


    
```
